# Group 3 — Frontend for the live demo

Keep this notebook next to the folder **`frontend`**, inside your group3-market folder (a subfolder is fine).

**Always open the page through cell 2** (`http://localhost:8000`). Double-clicking `index.html` opens it as a file, and MetaMask does not work there. Then run the cells from top to bottom.

1. **Names** — writes the six *public* addresses with your names into `frontend/members.js` (no private key leaves this cell).
2. **Start** — starts the website on `http://localhost:8000` and opens it.
3. **Stop** — stops it after the presentation.

Open the page in the browser where **MetaMask** is installed. When MetaMask asks which accounts to connect, tick **all six** (Platform, Max, Ben, Ale, Sarita, Anthony), so switching accounts during the demo works.

```python
# 1 · Names for the page: PUBLIC addresses only
import os, json
from dotenv import load_dotenv
from eth_account import Account

env_file = next((p for p in ["keys.env", ".env", "../keys.env", "../.env"] if os.path.exists(p)), None)
assert env_file, "keys.env not found - put this notebook in your group3-market folder (or one folder below it)"
load_dotenv(env_file)
print("Using", env_file)
def _key(v):
    v = (v or "").strip().strip('"').strip("'")
    return v if not v or v.startswith("0x") else "0x" + v

names = {}
k = _key(os.getenv("DEPLOYER_PRIVATE_KEY"))
if k: names[Account.from_key(k).address] = "Platform"
for i in range(1, 6):
    k = _key(os.getenv(f"MEMBER_{i}_PRIVATE_KEY"))
    if k: names[Account.from_key(k).address] = (os.getenv(f"MEMBER_{i}_NAME") or f"Member {i}").strip()

assert os.path.isdir("frontend"), "The folder 'frontend' must be in the same folder as this notebook"
with open("frontend/members.js", "w") as f:
    f.write("// Generated from keys.env - PUBLIC addresses only, safe to publish.\n"
            "window.TOBESURE_MEMBERS = " + json.dumps(names, indent=2) + ";\n")
for a, n in names.items():
    print(f"{n:<8} {a}")
print(f"\n✅ frontend/members.js written ({len(names)} wallets)")

# Safety check: the page must point at the same contract as the notebook
cfg = open("frontend/config.js").read()
addr = (os.getenv("CONTRACT_ADDRESS") or "").strip().lower()
print("✅ config.js uses your CONTRACT_ADDRESS" if addr and addr in cfg.lower() else
      "⚠️ CONTRACT_ADDRESS in keys.env differs from frontend/config.js - tell Claude")
```

```python
# 2 · Start the website and open it
import subprocess, sys, socket, time, webbrowser

def _free(port):
    with socket.socket() as s:
        return s.connect_ex(("127.0.0.1", port)) != 0

if "server" in globals() and server.poll() is None:
    print("Already running")
else:
    PORT = 8000
    while not _free(PORT):
        PORT += 1
    server = subprocess.Popen([sys.executable, "-m", "http.server", str(PORT), "--bind", "127.0.0.1", "--directory", "frontend"],
                              stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    time.sleep(1)
URL = f"http://localhost:{PORT}"
webbrowser.open(URL)
print("🌐", URL)
print("If it opened in a browser without MetaMask (e.g. Safari), copy this address into Chrome/Brave/Firefox.")
```

```python
# 3 · Stop the website (after the presentation)
if "server" in globals() and server.poll() is None:
    server.terminate(); print("Stopped")
else:
    print("Not running")
```
